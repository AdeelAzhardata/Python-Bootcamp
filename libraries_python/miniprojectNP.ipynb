```python
df = pd.read_csv("employee_data.csv")

df.head()
df.info()
df.describe()

df.isna().sum()

df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary']= df.groupby('Department')['Salary'].transform( lambda x: x.fillna(x.mean()))
df['PerformanceScore'].fillna(df['PerformanceScore'].mean() , inplace=True)

df.isna().sum()

df['Salary_Level'] = np.where(df['Salary']>70000 ,'High','Normal')

df.groupby('Department')['PerformanceScore'].mean()

df[df['PerformanceScore']>85]

df[['Experience','Salary']].corr()

df.to_csv('Clean_Employee_Data.csv')
```
