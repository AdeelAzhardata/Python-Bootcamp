# Regression imputer

```python
import sklearn
import pandas as pd
```

```python
from sklearn.linear_model import LinearRegression

df=pd.read_csv("titanic.csv")
```

```python
df.columns
```

```python
#SPLIT
known = df[df['age'].notna()]
missing = df[df['age'].isna()]

#Train 
X_train = known[['fare','pclass']]
y_train = known[['age']]

model = LinearRegression()
model.fit(X_train , y_train)

predicted_age = model.predict(missing[['fare','pclass']])

df.loc[df['age'].isna() , 'age'] = predicted_age

```
