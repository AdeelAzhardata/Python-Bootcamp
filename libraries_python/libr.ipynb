NumPy Basics

```python
#ARRAY 
a = np.array([1,2,3])
a

#2D ARRAY ( MATRIX)
b = np.array([[1,2,3],[4,5,6]])
b

#Number od Dimensions
b.ndim , a.ndim

#ROWS AND COLUMNS
b.shape , a.shape

#Total Elements
a.size , b.size

#Data Type
a.dtype , b.dtype

#Indexing

#1D ARRAY
arr = np.array([10,20,30,40,50])
arr[2]
arr[-1]

#2D Array

arr = np.array([[20,30,40],[50,60,70]])
arr[1,2]

#Slicing

 #1D ARRAY
arr[1:4]

#2D ARRAY
arr[0:2, 1:3]


#BOOLEAN INDEXING

arr[arr>30]

#Fancy Indexing

arr[[1]]

# Array Operations
arr + 5
arr - 5
arr * 5
arr / 5
arr ** 2
```

```python
#Reshape
arr = np.arange(16)
arr.reshape(4,4)


#Flatten

arr = np.array([[20,30,40],[50,60,70]])
arr.flatten()

#Concatenate

arr1 = np.array([[20,30,40],[50,60,70]])
arr2 = np.array([[1,2,3],[4,5,6]])

np.concatenate([arr1,arr2] , axis=1) #Column wise
np.concatenate([arr1,arr2] , axis=0) #Row wise

#Stack

np.vstack([arr1,arr2])
np.hstack([arr1,arr2])

#NumPy Random

np.random.rand(3,3)
np.random.randn(3,3)
np.random.randint(1,10,(3,3))

#arange and linspace

np.arange(0,10,2) #Step = 2
np.linspace(0,10,5) #to get Equal Points


#Broadcasting
b = np.array([[20],[30],[40]])
a = np.array([1,2,3])
a+b

#Aggregation Functions

marks = np.array([16,16,20,40,40,60,80,55,55])
marks.sum()
marks.mean()
marks.min()
marks.max()
marks.std()
marks.var()

#Mathematical Operations

np.sqrt(marks)
np.exp(marks)
np.log(marks)
np.unique(marks)




```
