```python
print("Smart Expense Checker")

def convert_to_numbers(exp_list):
    cleaned=[]
    for items in exp_list:
        if items.isdigit():
            cleaned.append(int(items))
        else:
            print(f"ignoring invalid input:{items}")
            continue
    return cleaned

def financial_advice(total):
    return (lambda x : "Great ! you managed your expenses well" if x < 2000 else "You spent alot today")(total)

try:
    input_raw_data = input("Enter your expense seperated by space:")

    expense_items = input_raw_data.split()

    expense = convert_to_numbers(expense_items)

    if not expense:
        print("No valid expense entered")
        pass
    else:
        total = sum(expense)
        highest = max(expense)
        lowest = min(expense)

        print("-----Expense Summary-----")

        print(f"Valid Expense:{expense}")
        print(f"Total expense:{total}")
        print(f"Highest expense:{highest}")
        print(f"Lowest Expense:{lowest}")


        print("Advice:",financial_advice(total))
except Exception as e:
    print("Error occured",e)

finally:
    print("Thankyou for using Smart expense Checker :) :) ")



```

"Smart to do Organizer" Programe



```python
print("----To Do Organizer----")


def clean_items(entry_items):

    cleaned=[]
    for items in entry_items:
        if all(char.isalnum() or char.isspace() for char in items):
            cleaned.append(items)
        else:
            print(f"Invalid entry :{items}")
            continue
    return cleaned
```

```python
print("Welcome to smart TO DO organizer")

def clean_task(task_list):
    cleaned=[]
    for t in task_list:
        task = t.strip()
        if task=="":
            continue
        cleaned.append(task)
        return cleaned
    
def categorize(task):
    task_lower=task.lower()
    if "work " in task_lower:
        return "Work"
    elif "study" in task_lower:
        return "Study"
    elif "home" in task_lower:
        return "Home"
    else:
        return "General"


try:
    input_raw_task=input("Enter tasks seperated by space")
    items = input_raw_task.split(",")
    tasks = clean_task(items)

    if not tasks:
        print("No valid task entered")
        pass

    else:

        categorized = [(t , categorize(t)) for t in tasks]

        priority_sort = lambda x:("!" in x[0],x[0])

        categorized = sorted (categorized , key = priority_sort , reverse=True)

        print("---Task summary---")
        for t ,cat in categorized:
            print(f"{t} ---->  {cat}")

        print(f"Total tasks: {len(tasks)}")

except Exception as e:
    print("Error ocurred:" ,e)

finally:
    print("Thankyou for using Smart TO DO organizer :) :) ")
```

Report card

```python
students=[]

def add_student(students):
    name=input("Enter Student's Name")
    marks=int(input("Enter Student marks from 0-100"))
    students.append({"name":name , "marks":marks})
    print("Student Added !! ")

def display_student(students):
    if not students:
        print("No student is added")
        return
    
    print("----Students List----")
    for s in students:
        print(f"{s["name"]} : {s["marks"]}")
    print()

def show_stats(students):
    if not students:
        print("No student to show")
        return
    
    total =0
    highest = students[0]
    lowest = students[0]

    for s in students:
        total+=s["marks"]

        if s["marks"]>highest["marks"]:
            highest=s
        if s["marks"]<lowest["marks"]:
            lowest=s

    avg = total/len(students)

    print("-----Stats of Students-----")

    print("Highest marks" ,  highest["name"] , highest["marks"])
    print("lowest marks" ,  lowest["name"] , lowest["marks"])
    print(f"Average Marks of batch:{avg}")
    print()

while True:
    print("Press 1 : Add Students data")
    print("Press 2 : Display students mark list")
    print("Press 3 : See Stats of batch")
    print("Press 4 : Exit")

    choice=input("Enter choice")

    if choice=="1":
       add_student(students)
    elif choice=="2":
        display_student(students)
    elif choice =="3":
       show_stats(students)
    elif choice =="4":
       print("Goodbye")
       break
    else:
     print("Invalid choice")

```
