```python
import pandas as pd
import sklearn 
```

```python
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv("titanic.csv")
```

```python
df.isna().sum().sort_values(ascending=False)
```

```python
num_col = df.select_dtypes(include=['int64','float64']).columns
```

# KNN Imputer

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df[num_col])
```

```python
imputer =   KNNImputer(n_neighbors=3,weights='distance')
imputed_scaled_data = imputer.fit_transform(scaled_data)
imputed_data = scaler.inverse_transform(imputed_scaled_data)
df[num_col]=pd.DataFrame(imputed_data, columns=num_col)
```

```python
df.isna().sum()
```

# Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['sex']= le.fit_transform(df['sex'])
df['sex']
```

# Iterative Imputer

```python
imputer = IterativeImputer(max_iter=10,random_state=42)
df[num_col] = imputer.fit_transform(df[num_col])
```

# Regression Imputer

```python
known = df[df['age'].notna()]
missing = df[df['age'].isna()]

X_train = known[['fare','pclass']]
y_train = known[['age']]

model = LinearRegression()
model.fit(X_train , y_train)

predicted_age = model.predict(missing[['fare','pclass']])
df.loc[df['age'].isna() , 'age'] = predicted_age
```

Cleaning Categorical data

```python
# df = df.drop('deck',axis=1)
df['embarked'].fillna(df['embarked'].mode()[0] , inplace=True)
df['embark_town'].fillna(df['embark_town'].mode()[0] , inplace=True)
df.isna().sum()
```
